```python
import tabula
import pandas as pd

# Extract tables from the PDF and convert them into a CSV file because with tabula it takes long time to interact with data.
tabula.convert_into(
    "Global_Urban_Temperatures.csv.pdf",
    "Global_Urban_Temperatures_extracted.csv",
    output_format="csv",
    pages="all",
    stream=True
)

```

```python
import numpy as np
import csv
rows = []

with open('Global_Urban_Temperatures_extracted.csv', newline="", encoding="utf-8", errors="replace") as f:
    reader = csv.reader(f)

    for row in reader:
        if not row:
            continue

        # If already 7 columns -> keep
        if len(row) == 7:
            rows.append(row)

        # If more than 7 -> city had commas, merge city parts back
        elif len(row) > 7:
            fixed = (
                row[:3] +                        # first 3 columns
                [",".join(row[3:-3])] +          # merge city parts
                [row[-3]] +                      # country
                row[-2:]                         # last 2 numeric columns
            )
            rows.append(fixed)

        # If fewer than 7 -> skip or pad (rare)
        else:
            continue

df = pd.DataFrame(rows, columns=[
    "Date", "AverageTemperature", "Value", "City", "Country", "TMAX", "TMIN"
])

# Convert missing values to NA
df = df.replace(-9999, pd.NA)

# convert missing AverageTemprature values to average of TMAX and TMIN
df["AverageTemperature"] = np.mean(df[["TMAX", "TMIN"]].astype(float), axis=1)



df.head()
```
