1. **Create a Raster:**
   - Create a 100x100 NumPy array to simulate a raster image with random integer values between 0 and 255 (representing grayscale pixel values).

```python
# Run Code here
import numpy as np
raster_image = np.random.randint(0, 256, size=(100, 100), dtype=np.uint8)
# Display the raster image using Matplotlib's imshow function with a grayscale colormap.
import matplotlib.pyplot as plt
plt.imshow(raster_image, cmap='gray')
plt.title('Random Grayscale Raster Image')
plt.show()
# Save the displayed raster image as a PNG file named 'random_raster_image.png'.
plt.imsave('random_raster_image.png', raster_image, cmap='gray')    
```

2.
   - Calculate the minimum, maximum, and mean values of the raster.
   - Count how many pixels have a value greater than 200.

```python
# Run Code here
# Calculate and print the following statistics of the raster image:
min_value = np.min(raster_image)
max_value = np.max(raster_image)
mean_value = np.mean(raster_image)
count_greater_200 = np.sum(raster_image > 200) 
```

3. **Modification:**
   - Increase the brightness of the raster by adding 50 to each pixel value, ensuring that the maximum value does not exceed 255.
    - Set all pixels with values below 50 to zero (simulate a threshold operation).

```python
# Run Code here
# Create a new raster image by increasing the brightness of the original raster image by 50 units.
brightened_raster = np.clip(raster_image + 50, 0, 255)
brightened_raster[brightened_raster < 50] = 0 
# Display the modified raster image using Matplotlib's imshow function with a grayscale colormap.
plt.imshow(brightened_raster, cmap='gray')
plt.title('Brightened Raster Image with Thresholding')
plt.show()
```
